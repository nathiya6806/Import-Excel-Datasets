```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("All libraries imported successfully!")
```

```python
df = pd.read_csv("StudentsPerformance.csv")

df.head()
```

```python
# Convert CSV dataset into Excel format

df.to_excel("StudentsPerformance.xlsx", index=False)

print("Excel file created successfully!")
```

```python
# Import Excel dataset

excel_df = pd.read_excel("StudentsPerformance.xlsx")

print("Excel dataset imported successfully!")

excel_df.head()
```

```python
print("Dataset Information:")

excel_df.info()
```

```python

# Check missing values
print("Missing Values:")
print(excel_df.isnull().sum())

# Check duplicate records
print("\nDuplicate Records:")
print(excel_df.duplicated().sum())

# Dataset shape
print("\nDataset Shape:")
print(excel_df.shape)

# Display column names
print("\nColumn Names:")
print(excel_df.columns.tolist())
```

```python

# Statistical summary
print("Statistical Analysis:")
display(excel_df.describe())

# Calculate average scores
import numpy as np

excel_df["average_score"] = np.mean(
    excel_df[["math score", "reading score", "writing score"]],
    axis=1
)

print("\nAverage Student Score:")
print(excel_df["average_score"].mean())

# Highest and lowest average scores
print("\nHighest Average Score:")
print(excel_df["average_score"].max())

print("\nLowest Average Score:")
print(excel_df["average_score"].min())
```

```python

import matplotlib.pyplot as plt

subjects = ["Math", "Reading", "Writing"]

average_scores = [
    excel_df["math score"].mean(),
    excel_df["reading score"].mean(),
    excel_df["writing score"].mean()
]

plt.figure(figsize=(8, 5))
plt.bar(subjects, average_scores)

plt.title("Average Student Performance by Subject")
plt.xlabel("Subjects")
plt.ylabel("Average Score")
plt.ylim(0, 100)

plt.show()
```

```python

test_prep_scores = excel_df.groupby(
    "test preparation course"
)[["math score", "reading score", "writing score"]].mean()

test_prep_scores.plot(
    kind="bar",
    figsize=(9, 5)
)

plt.title("Student Performance Based on Test Preparation")
plt.xlabel("Test Preparation Course")
plt.ylabel("Average Score")
plt.xticks(rotation=0)
plt.legend(title="Subjects")
plt.show()
```

```python

print("KEY FINDINGS")

print("1. Total number of students:", len(excel_df))

print("2. Average Math Score:",
      round(excel_df["math score"].mean(), 2))

print("3. Average Reading Score:",
      round(excel_df["reading score"].mean(), 2))

print("4. Average Writing Score:",
      round(excel_df["writing score"].mean(), 2))

print("\n5. Performance based on test preparation:")
print(test_prep_scores.round(2))

print("\n6. Conclusion:")
print("Reading and writing scores can be compared with math scores.")
print("The charts help understand student performance.")
```
